# Atividade 3 — Tabela 3 (IBGE): áreas gerais de formação

Pessoas com nível superior completo, por região, segundo o sexo, divididas em 3 tabelas:

1. **Total** de pessoas com curso superior, por região e sexo
2. **Ciência, Tecnologia, Engenharias e Matemática**, por região e sexo
3. **Educação, Serviços pessoais, Saúde e Bem-estar**, por região e sexo

Fonte: IBGE — Censo Demográfico (Tabela 10063 / SIDRA).

```python
import pandas as pd
```

## 1. Carregando a Tabela 3

O arquivo `Tabela_3_Areas_gerais_de_formacao_na_graduacao.xlsx` precisa estar na mesma pasta deste notebook.

```python
df = pd.read_excel("Tabela_3_Areas_gerais_de_formacao_na_graduacao.xlsx")
df.head()
```

## 2. Dividindo a tabela em 3 tabelas (homens e mulheres por região)

```python
curso_superior = df[["Região", "Total_Geral_Total", "Total_Geral_Homens", "Total_Geral_Mulheres"]].copy()
curso_superior.columns = ["Região", "Total", "Homens", "Mulheres"]
curso_superior
```

```python
tecnologia = df[["Região", "CTEM_Total", "CTEM_Homens", "CTEM_Mulheres"]].copy()
tecnologia.columns = ["Região", "Total", "Homens", "Mulheres"]
tecnologia
```

```python
educacao_saude = df[["Região", "ESSB_Total", "ESSB_Homens", "ESSB_Mulheres"]].copy()
educacao_saude.columns = ["Região", "Total", "Homens", "Mulheres"]
educacao_saude
```

## 3. Salvando as 3 tabelas em xlsx

```python
curso_superior.to_excel("curso_superior.xlsx", index=False)
tecnologia.to_excel("tecnologia.xlsx", index=False)
educacao_saude.to_excel("educacao_saude.xlsx", index=False)
```

## 4. Estatísticas descritivas (`describe()`) das 3 tabelas

```python
print("Curso superior (total geral):")
curso_superior.describe()
```

```python
print("Ciência, Tecnologia, Engenharias e Matemática:")
tecnologia.describe()
```

```python
print("Educação, Serviços pessoais, Saúde e Bem-estar:")
educacao_saude.describe()
```
