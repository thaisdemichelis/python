# Projeto: População por Município e Estado (Censo 2010 x 2022)

Fonte: IBGE — Censo Demográfico 2022, tabela de população compatibilizada com o Censo 2010.

Este notebook:
1. Lê a planilha original (`.xlsx`)
2. Agrega a população por **estado**
3. Ordena os estados pelo crescimento populacional (2022 - 2010)
4. Salva o resultado em CSV
5. Repete o processo por **município**


## 1. Importar bibliotecas

```python
import pandas as pd
```

## 2. Ler a tabela

A planilha tem 3 linhas de título/cabeçalho antes dos nomes reais das colunas (a linha de cabeçalho útil é a de índice 2, começando do zero) e a coluna A está vazia. Também há linhas de nota/fonte no final do arquivo, que serão descartadas.

```python
df = pd.read_excel(
    "data/CD2022_Populacao_2010_Compatibilizada_20231222.xlsx",
    sheet_name="Municípios",
    header=2,      
    usecols="B:H",  
)


df = df.dropna(subset=["COD. UF"]).copy()


df = df.rename(columns={
    "População Município 2010\n(Sinopse)": "Pop_2010_Sinopse",
    "População 2010 (Alterações de Limites até 2022)1": "Pop_2010_Compatibilizada",
    "População Censo 2022": "Pop_2022",
})

df["COD. UF"] = df["COD. UF"].astype(int)
df["COD. MUNIC"] = df["COD. MUNIC"].astype(int)

print(df.shape)
df.head()
```

## 3. Tabela agregada por ESTADO

Usamos `groupby` para somar a população de todos os municípios de cada UF, tanto para 2010 (base compatibilizada) quanto para 2022.

```python
pop_estado = (
    df.groupby("UF")[["Pop_2010_Compatibilizada", "Pop_2022"]]
    .sum()
    .reset_index()
)

pop_estado.head()
```

## 4. Ordenar por população que mais cresceu (2010 → 2022)

Calculamos a diferença absoluta entre 2022 e 2010, e também o crescimento percentual, e ordenamos do estado que mais cresceu para o que menos cresceu.

```python
pop_estado["Diferenca_2010_2022"] = pop_estado["Pop_2022"] - pop_estado["Pop_2010_Compatibilizada"]
pop_estado["Crescimento_%"] = (
    pop_estado["Diferenca_2010_2022"] / pop_estado["Pop_2010_Compatibilizada"] * 100
).round(2)

pop_estado = pop_estado.sort_values("Diferenca_2010_2022", ascending=False).reset_index(drop=True)

pop_estado
```

## 5. Salvar a tabela por estado em CSV

```python
pop_estado.to_csv(
    "output/populacao_por_estado.csv",
    sep=";",
    index=False,
    encoding="utf-8-sig",
)
print("Salvo em output/populacao_por_estado.csv")
```

## 6. Tabela por MUNICÍPIO

Aqui não é necessário agregar (cada linha já é um município), apenas calcular a diferença de população e ordenar.

```python
pop_municipio = df[[
    "UF", "COD. MUNIC", "NOME DO MUNICÍPIO",
    "Pop_2010_Compatibilizada", "Pop_2022",
]].copy()

pop_municipio["Diferenca_2010_2022"] = (
    pop_municipio["Pop_2022"] - pop_municipio["Pop_2010_Compatibilizada"]
)
pop_municipio["Crescimento_%"] = (
    pop_municipio["Diferenca_2010_2022"] / pop_municipio["Pop_2010_Compatibilizada"] * 100
).round(2)

pop_municipio = pop_municipio.sort_values(
    "Diferenca_2010_2022", ascending=False
).reset_index(drop=True)

pop_municipio.head(15)
```

## 7. Salvar a tabela por município em CSV

```python
pop_municipio.to_csv(
    "output/populacao_por_municipio.csv",
    sep=";",
    index=False,
    encoding="utf-8-sig",
)
print("Salvo em output/populacao_por_municipio.csv")
```
